# Welcome to The Virtual Brain — JupyterLite Lab

This is an interactive, browser-based **JupyterLab** environment running Python 3.13 via WebAssembly (`xeus-python`).
All of the documentation material is available in the **File Browser on the left**, and every notebook is live and executable directly in your browser — no server or local installation required.

## How to navigate and run code

- **Open any notebook** from the folders on the left (`tutorials/`, `howtos/`, `explanation/`, `demos/`, `lab/`).
- **Execute cells** with `Shift+Enter`, use **Run → Run All Cells**, or press `Shift+E` to restart the kernel and run an entire notebook.
- **Edit and save**: changes are stored in your browser's local storage. Use the **Download** toolbar button (`File → Download`) to export a modified notebook, or `Help → Clear Browser Data` to reset to the published files.
- **Launcher & Terminal**: click the `+` button at the top of the File Browser to open a new notebook, Python console, or WebAssembly terminal.
- **MCP Hub (optional)**: the **MCP Hub** panel is collapsed in the right sidebar rail if you want to pair this tab with an external MCP client.
